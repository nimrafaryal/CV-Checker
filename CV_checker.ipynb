```python
import fitz  # PyMuPDF

# Get uploaded file name
pdf_file = list(uploaded.keys())[0]

# Open PDF
doc = fitz.open(pdf_file)

# Extract text
text = ""

for page in doc:
    text += page.get_text()

print("===== EXTRACTED TEXT =====\n")
print(text)

# Clean text (remove extra spaces, line breaks)
clean_text = " ".join(text.split())

print("\n===== CLEANED TEXT PREVIEW =====\n")
print(clean_text[:2000])  # preview only

print("\n===== BASIC STATS =====")
print("Total characters:", len(clean_text))
print("Total words:", len(clean_text.split()))
```

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
job_description = """
We are looking for a Data Analyst with experience in SQL, Python, Power BI, and data visualization.
The candidate should be able to analyze datasets and generate insights.
"""
```

```python
resume_embedding = model.encode(clean_text, convert_to_tensor=True)
job_embedding = model.encode(job_description, convert_to_tensor=True)
```

```python
score = util.pytorch_cos_sim(resume_embedding, job_embedding)

print("===== MATCH RESULT =====")
print("Similarity Score:", round(score.item() * 100, 2), "%")
```

```python
percent = round(score.item() * 100, 2)

print("\n===== AI RESUME ANALYSIS =====")

if percent > 75:
    print("🟢 Strong Match")
elif percent > 50:
    print("🟡 Moderate Match")
else:
    print("🔴 Weak Match")

print(f"Match Score: {percent}%")
```

```python
def generate_ai_feedback(resume_text, job_description, score):

    prompt = f"""
You are an expert ATS recruiter and career coach.

Analyze the resume against the job description.

Return structured feedback:

1. ATS Score: {score}%
2. Missing Skills
3. Weak Areas in Resume
4. Suggestions to improve resume
5. Career Advice

---

RESUME:
{resume_text}

---

JOB DESCRIPTION:
{job_description}
"""

    return prompt
```

```python
feedback = generate_ai_feedback(clean_text, job_description, round(score.item()*100,2))

print("===== AI FEEDBACK SYSTEM =====\n")
print(feedback)
```

```python
skill_list = [
    "SQL", "Python", "Power BI", "Excel",
    "Machine Learning", "Data Visualization",
    "Statistics", "Pandas"
]

missing_skills = []

for skill in skill_list:
    if skill.lower() not in clean_text.lower():
        missing_skills.append(skill)
```

```python
print("\n===== FINAL AI REPORT =====")

score_percent = round(score.item() * 100, 2)

print(f"ATS Score: {score_percent}%")

print("\nMissing Skills:")
for s in missing_skills:
    print("-", s)

print("\nRecommendation:")

if score_percent >= 75:
    print("🟢 Strong match – You are highly recommended for this role.")
elif score_percent >= 50:
    print("🟡 Moderate match – You can apply but improve skills first.")
else:
    print("🔴 Weak match – Not recommended yet. Improve before applying.")

print("\n===== AI INSIGHTS =====")

if score_percent < 40:
    print("⚠️ Major skill gap detected. Focus on fundamentals first.")
elif score_percent < 70:
    print("📈 You are on the right track. Strengthen missing technical skills.")
else:
    print("🚀 Strong profile. Focus on projects and experience improvement.")

print("\n===== CAREER COACH ADVICE =====")

if "SQL" in missing_skills:
    print("- Learn SQL for data querying roles")

if "Power BI" in missing_skills:
    print("- Build 1 dashboard project in Power BI")

if "Python" in missing_skills:
    print("- Practice Python for data analysis (Pandas, NumPy)")
```

```python
%%writefile app.py
import streamlit as st
import fitz
from sentence_transformers import SentenceTransformer, util
import re

# ── Page config ──────────────────────────────────────────────────────────────
st.set_page_config(
    page_title="AI Resume Career Coach",
    page_icon="🧠",
    layout="wide",
)

# ── Custom CSS ────────────────────────────────────────────────────────────────
st.markdown("""
<style>
@import url('https://fonts.googleapis.com/css2?family=Syne:wght@400;600;700;800&family=DM+Sans:wght@300;400;500&display=swap');

html, body, [class*="css"] {
    font-family: 'DM Sans', sans-serif;
}
h1, h2, h3, .big-title {
    font-family: 'Syne', sans-serif !important;
}

.stApp {
    background: #0d0f14;
    color: #e8e8e8;
}

.score-card {
    background: linear-gradient(135deg, #1a1d26 0%, #1f2333 100%);
    border: 1px solid #2a2f45;
    border-radius: 16px;
    padding: 28px 32px;
    text-align: center;
    margin-bottom: 16px;
}
.score-number {
    font-family: 'Syne', sans-serif;
    font-size: 72px;
    font-weight: 800;
    line-height: 1;
    margin: 8px 0;
}
.score-label {
    font-size: 13px;
    letter-spacing: 2px;
    text-transform: uppercase;
    color: #6b7280;
}

.section-card {
    background: #13151e;
    border: 1px solid #1e2235;
    border-radius: 14px;
    padding: 22px 26px;
    margin-bottom: 14px;
}
.section-title {
    font-family: 'Syne', sans-serif;
    font-size: 13px;
    letter-spacing: 2px;
    text-transform: uppercase;
    color: #6b7280;
    margin-bottom: 14px;
}

.skill-pill {
    display: inline-block;
    background: #1e2235;
    border: 1px solid #2a3050;
    border-radius: 20px;
    padding: 5px 14px;
    margin: 4px;
    font-size: 13px;
    color: #a0aec0;
}
.skill-pill.missing {
    background: rgba(239,68,68,0.1);
    border-color: rgba(239,68,68,0.3);
    color: #f87171;
}
.skill-pill.present {
    background: rgba(52,211,153,0.1);
    border-color: rgba(52,211,153,0.3);
    color: #34d399;
}

.advice-item {
    display: flex;
    align-items: flex-start;
    gap: 12px;
    padding: 10px 0;
    border-bottom: 1px solid #1e2235;
    font-size: 14px;
    color: #cbd5e1;
    line-height: 1.6;
}
.advice-item:last-child { border-bottom: none; }
.advice-icon { font-size: 18px; flex-shrink: 0; margin-top: 1px; }

.stProgress > div > div > div > div {
    background: linear-gradient(90deg, #6366f1, #8b5cf6);
    border-radius: 99px;
}
.stProgress > div > div > div {
    background: #1e2235;
    border-radius: 99px;
}

div[data-testid="stFileUploader"] {
    background: #13151e;
    border: 1.5px dashed #2a2f45;
    border-radius: 12px;
    padding: 8px;
}
textarea {
    background: #13151e !important;
    border: 1px solid #2a2f45 !important;
    color: #e8e8e8 !important;
    border-radius: 10px !important;
}
.stButton > button {
    background: linear-gradient(135deg, #6366f1, #8b5cf6);
    color: white;
    border: none;
    border-radius: 10px;
    padding: 10px 28px;
    font-family: 'Syne', sans-serif;
    font-weight: 600;
    letter-spacing: 0.5px;
    transition: opacity 0.2s;
}
.stButton > button:hover { opacity: 0.85; }

.stTextArea label, .stFileUploader label {
    color: #94a3b8 !important;
    font-size: 13px !important;
    letter-spacing: 1px !important;
    text-transform: uppercase !important;
}
</style>
""", unsafe_allow_html=True)

# ── Model (cached) ────────────────────────────────────────────────────────────
@st.cache_resource
def load_model():
    return SentenceTransformer('all-MiniLM-L6-v2')

model = load_model()

# ── Helpers ───────────────────────────────────────────────────────────────────
def extract_text(pdf_file):
    doc = fitz.open(stream=pdf_file.read(), filetype="pdf")
    text = ""
    for page in doc:
        text += page.get_text()
    return " ".join(text.split())

def count_words(text):
    return len(text.split())

def extract_email(text):
    match = re.search(r'[\w.+-]+@[\w-]+\.[a-zA-Z]{2,}', text)
    return match.group(0) if match else None

def extract_phone(text):
    match = re.search(r'(\+?\d[\d\s\-]{8,}\d)', text)
    return match.group(0).strip() if match else None

def check_sections(text):
    keywords = {
        "Education":    ["education", "degree", "university", "bachelor", "master", "bs", "ms"],
        "Experience":   ["experience", "work", "employed", "company", "position", "role"],
        "Skills":       ["skills", "technical skills", "technologies", "proficient"],
        "Projects":     ["project", "built", "developed", "created"],
        "Summary":      ["summary", "objective", "profile", "about me"],
        "Certifications": ["certificate", "certification", "certified", "course"],
    }
    found, missing = [], []
    lower = text.lower()
    for section, kws in keywords.items():
        if any(k in lower for k in kws):
            found.append(section)
        else:
            missing.append(section)
    return found, missing

SKILL_ADVICE = {
    "SQL":               ("🗄️", "Practice SQL on Mode Analytics or LeetCode – focus on window functions, CTEs, and joins."),
    "Python":            ("🐍", "Work through Pandas & NumPy via Kaggle notebooks; aim for 2 end-to-end projects."),
    "Machine Learning":  ("🤖", "Take Andrew Ng's ML course on Coursera and implement 3 models from scratch."),
    "Statistics":        ("📐", "Review hypothesis testing, distributions, and Bayesian thinking via StatQuest on YouTube."),
    "Power BI":          ("📊", "Build a personal finance or sales dashboard and publish it to Power BI Service."),
    "Data Visualization":("🎨", "Learn Plotly & Seaborn; create a portfolio with 5 different chart types per project."),
    "Excel":             ("📑", "Master VLOOKUP, pivot tables, and Power Query through Excel practice files."),
    "Deep Learning":     ("🧠", "Complete fast.ai Part 1 and fine-tune a pre-trained model on a custom dataset."),
    "NLP":               ("💬", "Build a text classification project using HuggingFace Transformers."),
    "Tableau":           ("📈", "Connect Tableau Public to a real dataset and build an interactive story."),
    "R":                 ("📉", "Practice R for statistical modeling via R for Data Science (free online book)."),
    "TensorFlow":        ("⚙️",  "Follow the official TensorFlow tutorials and build a CNN image classifier."),
    "PyTorch":           ("🔥", "Implement a simple neural net from scratch using PyTorch tensors."),
    "Cloud (AWS/GCP/Azure)": ("☁️", "Get the AWS Cloud Practitioner or GCP Associate cert to validate cloud knowledge."),
}

ALL_SKILLS = list(SKILL_ADVICE.keys())

# ── UI ────────────────────────────────────────────────────────────────────────
st.markdown("""
<div style="padding: 32px 0 8px 0;">
  <div style="font-family:'Syne',sans-serif; font-size:36px; font-weight:800; color:#e2e8f0; line-height:1.1;">
    AI Resume<br><span style="color:#8b5cf6;">Career Coach</span>
  </div>
  <div style="color:#4b5563; font-size:14px; margin-top:8px; letter-spacing:0.5px;">
    Semantic ATS scoring · skill gap analysis · actionable advice
  </div>
</div>
<hr style="border-color:#1e2235; margin: 20px 0 28px 0;">
""", unsafe_allow_html=True)

col_left, col_right = st.columns([1, 1], gap="large")

with col_left:
    uploaded_file = st.file_uploader("Upload Resume (PDF)", type=["pdf"])
    job_description = st.text_area("Paste Job Description", height=260,
                                   placeholder="Paste the full job description here…")

with col_right:
    st.markdown("#### How it works")
    st.markdown("""
- 🔍 **Semantic similarity** — goes beyond keyword matching
- 🧩 **Skill gap detection** — flags what's missing vs. the JD
- 📋 **Resume health check** — section completeness score
- 💡 **Personalised advice** — specific next steps for each gap
- 📄 **Quick stats** — word count, contact info, key sections
""")

st.markdown("<br>", unsafe_allow_html=True)

# ── Analysis ──────────────────────────────────────────────────────────────────
if uploaded_file and job_description and len(job_description.strip()) > 30:

    with st.spinner("Analysing your resume…"):
        resume_text = extract_text(uploaded_file)

        resume_emb = model.encode(resume_text, convert_to_tensor=True)
        job_emb    = model.encode(job_description, convert_to_tensor=True)
        score = round(util.pytorch_cos_sim(resume_emb, job_emb).item() * 100, 1)

        missing_skills  = [s for s in ALL_SKILLS if s.lower() not in resume_text.lower()]
        present_skills  = [s for s in ALL_SKILLS if s.lower() in resume_text.lower()]
        found_secs, missing_secs = check_sections(resume_text)
        section_score = round(len(found_secs) / len(check_sections.__defaults__[0] if False else
                              ["Education","Experience","Skills","Projects","Summary","Certifications"]) * 100)
        # recalc cleanly
        total_secs = len(found_secs) + len(missing_secs)
        section_score = round(len(found_secs) / total_secs * 100) if total_secs else 0

        word_count = count_words(resume_text)
        email      = extract_email(resume_text)
        phone      = extract_phone(resume_text)

    # ── Score row ─────────────────────────────────────────────────────────────
    c1, c2, c3 = st.columns(3, gap="medium")

    score_color = "#34d399" if score >= 75 else "#fbbf24" if score >= 50 else "#f87171"
    match_label = "Strong Match" if score >= 75 else "Moderate Match" if score >= 50 else "Weak Match"
    match_emoji = "🚀" if score >= 75 else "🟡" if score >= 50 else "🔴"

    with c1:
        st.markdown(f"""
        <div class="score-card">
            <div class="score-label">ATS Score</div>
            <div class="score-number" style="color:{score_color};">{score}%</div>
            <div style="color:{score_color}; font-size:14px; font-weight:600;">{match_emoji} {match_label}</div>
        </div>""", unsafe_allow_html=True)

    with c2:
        st.markdown(f"""
        <div class="score-card">
            <div class="score-label">Resume Health</div>
            <div class="score-number" style="color:#6366f1;">{section_score}%</div>
            <div style="color:#6b7280; font-size:14px;">{len(found_secs)}/{total_secs} sections found</div>
        </div>""", unsafe_allow_html=True)

    with c3:
        wc_color = "#34d399" if 400 <= word_count <= 800 else "#fbbf24"
        st.markdown(f"""
        <div class="score-card">
            <div class="score-label">Word Count</div>
            <div class="score-number" style="color:{wc_color};">{word_count}</div>
            <div style="color:#6b7280; font-size:14px;">{'Ideal range ✓' if 400 <= word_count <= 800 else 'Aim for 400–800'}</div>
        </div>""", unsafe_allow_html=True)

    st.markdown("<br>", unsafe_allow_html=True)

    # ── Main two-column detail ─────────────────────────────────────────────────
    left, right = st.columns([1.1, 1], gap="large")

    with left:
        # Skill breakdown
        st.markdown('<div class="section-card">', unsafe_allow_html=True)
        st.markdown('<div class="section-title">Skill Breakdown</div>', unsafe_allow_html=True)

        if present_skills:
            st.markdown("**Detected in your resume:**")
            pills = "".join(f'<span class="skill-pill present">✓ {s}</span>' for s in present_skills)
            st.markdown(f'<div>{pills}</div>', unsafe_allow_html=True)

        if missing_skills:
            st.markdown("<br>**Not found / consider adding:**", unsafe_allow_html=True)
            pills = "".join(f'<span class="skill-pill missing">✗ {s}</span>' for s in missing_skills)
            st.markdown(f'<div>{pills}</div>', unsafe_allow_html=True)

        st.markdown('</div>', unsafe_allow_html=True)

        # Progress bars
        st.markdown('<div class="section-card">', unsafe_allow_html=True)
        st.markdown('<div class="section-title">Score Breakdown</div>', unsafe_allow_html=True)
        st.write("**Semantic match**")
        st.progress(int(score))
        st.write("**Resume completeness**")
        st.progress(section_score)
        skill_coverage = round(len(present_skills) / len(ALL_SKILLS) * 100) if ALL_SKILLS else 0
        st.write(f"**Skill coverage** ({len(present_skills)}/{len(ALL_SKILLS)})")
        st.progress(skill_coverage)
        st.markdown('</div>', unsafe_allow_html=True)

    with right:
        # Career advice (fixed + rich)
        st.markdown('<div class="section-card">', unsafe_allow_html=True)
        st.markdown('<div class="section-title">Career Advice</div>', unsafe_allow_html=True)

        items = []

        # General advice based on score
        if score >= 75:
            items.append(("🚀", "Strong match! Focus on tailoring your bullet points with <strong>quantified achievements</strong> (e.g. 'improved accuracy by 18%')."))
        elif score >= 50:
            items.append(("📈", "You're close to a strong match. Address the missing skills below and rewrite your summary to mirror the JD's language."))
        else:
            items.append(("⚠️", "Significant gap detected. Study the JD carefully and restructure your resume to highlight the most relevant experience first."))

        # Missing section advice
        if missing_secs:
            items.append(("📋", f"Add missing resume sections: <strong>{', '.join(missing_secs)}</strong> — recruiters and ATS scanners look for these explicitly."))

        # Word count advice
        if word_count < 300:
            items.append(("✍️", "Your resume is too short. Expand each role with 3–5 bullet points describing impact and tools used."))
        elif word_count > 900:
            items.append(("✂️", "Your resume may be too long. Trim to 1–2 pages; remove duties and keep only measurable achievements."))

        # Contact info
        if not email:
            items.append(("📧", "No email detected. Make sure your email address is clearly visible at the top."))
        if not phone:
            items.append(("📱", "No phone number detected. Add a contact number — many recruiters prefer a quick call."))

        # Skill-specific advice
        advice_count = 0
        for skill in missing_skills:
            if skill in SKILL_ADVICE and advice_count < 4:
                icon, text = SKILL_ADVICE[skill]
                items.append((icon, f"<strong>{skill}:</strong> {text}"))
                advice_count += 1

        # Generic tailoring tip
        items.append(("🎯", "Mirror exact keywords from the job description in your resume — ATS systems rank keyword density heavily."))

        for icon, text in items:
            st.markdown(f'<div class="advice-item"><span class="advice-icon">{icon}</span><span>{text}</span></div>', unsafe_allow_html=True)

        st.markdown('</div>', unsafe_allow_html=True)

    # ── Resume health detail ───────────────────────────────────────────────────
    st.markdown('<div class="section-card">', unsafe_allow_html=True)
    st.markdown('<div class="section-title">Resume Section Health</div>', unsafe_allow_html=True)

    cols = st.columns(6)
    all_sections = found_secs + missing_secs
    for i, sec in enumerate(sorted(set(all_sections))):
        found = sec in found_secs
        color = "#34d399" if found else "#f87171"
        mark  = "✓" if found else "✗"
        cols[i % 6].markdown(f"""
        <div style="text-align:center; padding:12px 6px; background:#1a1d26; border-radius:10px;
                    border:1px solid {'#1a4a38' if found else '#4a1a1a'}; margin:4px;">
            <div style="font-size:20px;">{mark}</div>
            <div style="font-size:11px; color:{color}; margin-top:4px;">{sec}</div>
        </div>""", unsafe_allow_html=True)

    st.markdown('</div>', unsafe_allow_html=True)

elif uploaded_file and not job_description:
    st.info("📋 Paste a job description on the left to start the analysis.")
elif not uploaded_file:
    st.markdown("""
    <div style="text-align:center; padding:60px 20px; color:#374151;">
        <div style="font-size:48px; margin-bottom:16px;">📄</div>
        <div style="font-family:'Syne',sans-serif; font-size:18px; color:#4b5563;">
            Upload your resume and paste a job description to begin
        </div>
    </div>
    """, unsafe_allow_html=True)
```
